```python
from typing import TypedDict, List
from functools import reduce
from operator import mul
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    operation: str
    result: str
```

```python
def process_values(state: AgentState) -> AgentState:
    """Process values based on the requested operation."""

    print("Input State:", state)

    if state["operation"] == "+":
        answer = sum(state["values"])
    
    elif state["operation"] == "*":
        answer = reduce(mul, state["values"], 1) # reduce(function, iterable, initial_value)
    
    else:
        answer = "Invalid operation."
    
    state["result"] = f"Hi {state['name']}! Your result = {answer}"

    print("Output State:", state)
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("processor", process_values)
graph.set_entry_point("processor")
graph.set_finish_point("processor")

app = graph.compile()
```

```python
# Invoke the graph

result = app.invoke({
    "values": [1,2,3,4],
    "name": "Jack Sparrow",
    "operation": "*"
}
)

```
