```python
from typing import TypedDict, List
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    age: int
    skills: List[str]
    result: str
```

```python
def greet(state: AgentState) -> AgentState:
    state["result"] = f"{state['name']}, welcome to the system."
    return state
```

```python
def add_age(state: AgentState) -> AgentState:
    # Only add the age once
    if "years old" not in state["result"]:
        state["result"] += f" You are {state['age']} years old."
    return state
```

```python
def add_skills(state: AgentState) -> AgentState:
    skills = ", ".join(state["skills"])
    state["result"] += f" You have skills in {skills}."
    return state
```

```python
graph = StateGraph(AgentState)

# graph.add_node(node_name, function)

graph.add_node("greet", greet)
graph.add_node("age1", add_age)
graph.add_node("age2", add_age)      # Same function used again
graph.add_node("skills", add_skills)

graph.set_entry_point("greet")

graph.add_edge("greet", "age1")
graph.add_edge("age1", "age2")
graph.add_edge("age2", "skills")

graph.set_finish_point("skills")

app = graph.compile()
```

```python
result = app.invoke(
    {
        "name": "Sam",
        "age": 31,
        "skills": ["Python", "Machine Learning", "LangGraph"],
    }
)
```
