```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    message: str

def compliment_node(state: AgentState) -> AgentState:
    """Adds a personalized compliment without replacing the existing state."""

    # Concatinating instead of replacing
    state["message"] = (
        state["message"] + ", you're doing an amazing job learing Langgraph!"
    
    )

    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("compliment", compliment_node)

graph.set_entry_point("compliment")
graph.set_finish_point("compliment")

app = graph.compile()
```

```python
result = app.invoke({"message": "Bob"})
print(result["message"])
```
